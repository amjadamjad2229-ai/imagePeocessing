```python
my_list = [1, 2, 3, 4, 5]
print(f"Original list: {my_list}")
my_list.append(6)
print(f"List after append: {my_list}")
print("\nLooping through list:")
for item in my_list:
    print(item * 2)
def calculate_square(number):
    return number ** 2
print(f"\nSquare of 5: {calculate_square(5)}")
def process_list(input_list):
    processed = []
    for num in input_list:
        processed.append(calculate_square(num))
    return processed
print(f"Processed list (squares): {process_list(my_list)}")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
opencv_image_path = 'sample_image_cv.jpg'
pil_image_path = 'sample_image_pil.png'
print(f"Attempting to load image with OpenCV from: {opencv_image_path}")
if os.path.exists(opencv_image_path):
    img_cv = cv2.imread(opencv_image_path)
    if img_cv is None:
        print(f"Warning: Could not read {opencv_image_path} with OpenCV. It might be corrupted or an unsupported format. Generating dummy image instead.")
        img_cv = np.zeros((300, 400, 3), dtype=np.uint8)
        cv2.putText(img_cv, "Dummy OpenCV Image", (50, 150), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
else:
    print(f"File not found: {opencv_image_path}. Generating dummy image with NumPy.")
    img_cv = np.zeros((300, 400, 3), dtype=np.uint8)
    cv2.putText(img_cv, "Dummy OpenCV Image", (50, 150), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
img_cv_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_cv_rgb)
plt.title('Image loaded with OpenCV (RGB)')
plt.axis('off')
print(f"\nAttempting to load image with PIL from: {pil_image_path}")
if os.path.exists(pil_image_path):
    try:
        img_pil = Image.open(pil_image_path)
    except Exception as e:
        print(f"Warning: Could not read {pil_image_path} with PIL due to: {e}. Generating dummy image instead.")
        img_pil = Image.fromarray(np.uint8(np.random.rand(300, 400) * 255))
        img_pil = img_pil.convert('L')
else:
    print(f"File not found: {pil_image_path}. Generating dummy image with NumPy and PIL.")
    dummy_array = np.uint8(np.random.rand(300, 400) * 255)
    img_pil = Image.fromarray(dummy_array, mode='L')
plt.subplot(1, 2, 2)
plt.imshow(np.array(img_pil), cmap='gray')
plt.title('Image loaded with PIL (Grayscale)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
new_opencv_filename = 'new_image_cv.jpg'
new_pil_filename = 'new_image_pil.jpg'
img_cv_bgr = cv2.cvtColor(img_cv_rgb, cv2.COLOR_RGB2BGR)
cv2.imwrite(new_opencv_filename, img_cv_bgr)
print(f"OpenCV image saved as: {new_opencv_filename}")
img_pil.save(new_pil_filename)
print(f"PIL image saved as: {new_pil_filename}")
```

```python
import matplotlib.pyplot as plt
height, width, _ = img_cv_rgb.shape
startY, endY = int(height * 0.25), int(height * 0.75)
startX, endX = int(width * 0.25), int(width * 0.75)
roi = img_cv_rgb[startY:endY, startX:endX]
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_cv_rgb)
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(roi)
plt.title('Region of Interest (ROI)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
modified_img = img_cv_rgb.copy()
mod_startY, mod_endY = int(height * 0.3), int(height * 0.7)
mod_startX, mod_endX = int(width * 0.3), int(width * 0.7)
modified_img[mod_startY:mod_endY, mod_startX:mod_endX] = np.clip(
    modified_img[mod_startY:mod_endY, mod_startX:mod_endX] + 80, 0, 255
)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(img_cv_rgb)
plt.title('Original Image')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(modified_img)
plt.title('Image with Brighter Region')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
if img_cv_rgb.ndim == 2:
    print("Original image is grayscale. Creating a synthetic RGB image for channel splitting.")
    rgb_image_for_channels = np.stack([img_cv_rgb, img_cv_rgb, img_cv_rgb], axis=-1)
elif img_cv_rgb.shape[2] == 1:
    print("Original image is 1-channel. Creating a synthetic RGB image for channel splitting.")
    rgb_image_for_channels = np.stack([img_cv_rgb.squeeze(), img_cv_rgb.squeeze(), img_cv_rgb.squeeze()], axis=-1)
else:
    rgb_image_for_channels = img_cv_rgb.copy()
R, G, B = rgb_image_for_channels[:, :, 0], rgb_image_for_channels[:, :, 1], rgb_image_for_channels[:, :, 2]
plt.figure(figsize=(15, 5))
plt.subplot(1, 4, 1)
plt.imshow(rgb_image_for_channels)
plt.title('Original RGB')
plt.axis('off')
plt.subplot(1, 4, 2)
plt.imshow(R, cmap='gray')
plt.title('Red Channel')
plt.axis('off')
plt.subplot(1, 4, 3)
plt.imshow(G, cmap='gray')
plt.title('Green Channel')
plt.axis('off')
plt.subplot(1, 4, 4)
plt.imshow(B, cmap='gray')
plt.title('Blue Channel')
plt.axis('off')
plt.tight_layout()
plt.show()
zeros = np.zeros_like(G)
green_only_image = np.stack([zeros, G, zeros], axis=-1)
swapped_rb_image = np.stack([B, G, R], axis=-1)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(green_only_image)
plt.title('Green Channel Only')
plt.axis('off')
plt.subplot(1, 2, 2)
plt.imshow(swapped_rb_image)
plt.title('Red & Blue Channels Swapped')
plt.axis('off')
plt.tight_layout()
plt.show()
```
